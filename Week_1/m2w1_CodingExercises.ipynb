```python
#Exercise Set 3: GroupBy and Aggregation
```

```python
import pandas as pd

data = {
"Store": ["A", "A", "A", "B", "B", "C", "C", "C", "C"],
"Product": ["Apples", "Oranges", "Bananas", "Apples", "Bananas", "Oranges", "Apples", "Bananas", "Bananas"],
"Sales": [100, 80, 90, 120, 110, 60, 95, 85, 75],
"Profit": [30, 25, 20, 40, 35, 15, 32, 22, 18],
}

df = pd.DataFrame(data)
print(df.head())
```

```python
# Exercise 1 Group the data by Store and use .agg() to find the average Sales and average Profit for each store.
result = df.groupby('Store').agg(
    Avg_Sales=('Sales', 'mean'),
    Avg_Profit=('Profit', 'mean')
).reset_index()



print(result.to_string(index=False))
```

```python
#Exercise 2 Group by Product and get the maximum Sales and minimum Profit.
result = df.groupby('Product').agg(
    Sales=('Sales', 'max'),
    Profit=('Profit', 'min')
).reset_index()

print(result.to_string(index=False))
```

```python
#Exercise 3 Group by Store and return the average Sales and maximum Profit, but name the columns 
# Avg_Sales and Max_Profit.

result = df.groupby('Store').agg(
    Ave_Sales=('Sales', 'mean'),
    Max_Profit=('Profit', 'max')
).reset_index()

print(result.to_string(index=False))
```

```python
#Exercise 4 BONUS: Group by both Store and Product, and find the total Sales using .agg().
result = df.groupby(['Store','Product']).agg(
    Total_Sales=('Sales', 'sum')
    
).reset_index()
print(result.to_string(index=False))
```
